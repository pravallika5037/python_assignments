```python
#Question 1: Create a set of first 10 natural numbers
#solution :
numbers=set(range(1,11))
print("Set:",numbers)
```

```python
#Question 2: Add element 11 to set {1, 2, 3, 4, 5}
#solution :
numbers={1,2,3,4,5}
numbers.add(11)
print("Set:",numbers)
```

```python
#Question 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}
#solution :
numbers={1,2,3,4,5,6}
numbers.remove(3)
print("Set:",numbers)
```

```python
#Question 4: Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
#solution :
set1={1,2,3,4,5}
set2={4,5,6,7,8}
result=set1.intersection(set2)
print("Intersection:",result)
```

```python
#Question 5: Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
#solution :
set1={1,2,3,4,5}
set2={4,5,6,7,8}
result=set1.difference(set2)
print("Difference:",result)
```

```python
#Question 6: Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}
#solution :
numbers={1,2,3,4,6,7,8}
print("5 is present:",5 in numbers)
```

```python
#Question 7: Find the length of set {'a', 'b', 'c', 'd', 'e'}
#solution :
letters={"a","b","c","d","e"}
print("Length:",len(letters))
```

```python
#Question 8: Create a set of vowels from string 'hello world'
#solution :
text="hello world"
vowels=set()
for char in text:
    if char in "aeiou":
        vowels.add(char)
print("Vowels:",vowels)
```

```python
#Question 9: Remove duplicates from list [1, 2, 2, 3, 4, 4, 5, 6, 6, 7] using set
#solution :
numbers=[1,2,2,3,4,4,5,6,6,7]
result=set(numbers)
print("Without duplicates:",result)
```

```python
#Question 10: Check if {1, 2, 3} is a subset of {1, 2, 3, 4, 5, 6}
#solution :
set1={1,2,3}
set2={1,2,3,4,5,6}
print("Is subset:",set1.issubset(set2))
```
