````python
n = int(input("Enter a number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome number")
else:
    print("Not a palindrome number")
````

Problem 3.5: Armstrong Number (Hard)
Write a program to check if a number is an Armstrong number (also called Narcissistic number).

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

Examples:

153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 (Armstrong number)
9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)
Hint: First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

````python
n = int(input("Enter a number: "))

original = n
power = len(str(n))
total = 0

while n > 0:
    digit = n % 10
    total = total + digit ** power
    n = n // 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
````

Problem 6.2: GCD Calculator (Medium)
Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

Algorithm:

Divide the larger number by the smaller number
Replace the larger number with the smaller number
Replace the smaller number with the remainder
Repeat until the remainder is 0
The GCD is the last non-zero remainder
Hint: While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

````python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD =", a)
````

Problem 6.5: Number to Words (Hard)
Write a program that converts a number (0-999) to words.

Examples:

0: Zero
25: Twenty Five
123: One Hundred Twenty Three
500: Five Hundred
Hint: Create lists for ones, teens, tens, etc. Extract hundreds, tens, ones digits. Convert each part to words and combine.

````python
ones = ["", "One", "Two", "Three", "Four", "Five",
        "Six", "Seven", "Eight", "Nine"]

teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen",
         "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"]

tens = ["", "", "Twenty", "Thirty", "Forty",
        "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"]

n = int(input("Enter a number (0-999): "))

if n == 0:
    print("Zero")
elif n < 0 or n > 999:
    print("Enter a number between 0 and 999")
else:
    words = ""

    # Extract hundreds digit
    hundreds = n // 100
    if hundreds > 0:
        words += ones[hundreds] + " Hundred "
        n = n % 100
    # Convert tens and ones
    if n >= 10 and n <= 19:
        words += teens[n - 10]
    else:
        t = n // 10
        o = n % 10
        words += tens[t]

        if o > 0:
            words += " " + ones[o]

    print(words.strip())
````

Problem 6.6: Strong Number Checker (Hard)
Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

Hint: Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

````python
n = int(input("Enter a number: "))

original = n
total = 0

while n > 0:
    digit = n % 10

    # Calculate factorial of each digit
    fact = 1
    for i in range(1, digit + 1):
        fact *= i

    total += fact
    n //= 10

if total == original:
    print(original, "is a Strong Number")
else:
    print(original, "is not a Strong Number")
````

### Problem 6.7: Pascal's Triangle (Hard)

Write a program to print Pascal's Triangle up to n rows.

Example for n=5:
```
    1
   1 1
  1 2 1
 1 3 3 1
1 4 6 4 1
```

Each number is the sum of the two numbers directly above it.

**Hint:** Use formula: C(n,k) = C(n,k-1) * (n-k+1) / k, where C(n,0)=1. Or use nested loops with list to store previous row.

````python
n = int(input("Enter number of rows: "))

for i in range(n):
    c = 1

    # Print spaces for triangle shape
    print(" " * (n - i), end="")

    for k in range(1, i + 2):
        print(c, end=" ")
        c = c * (i - k + 1) // k

    print()
````
