Write a program to check if a number is an Armstrong number (also called Narcissistic number).

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

Problem 6.6: Strong Number Checker (Hard)
Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

Hint: Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

````python
n = int(input("Enter a number: "))

original = n
total = 0

while n > 0:
    digit = n % 10

    # Calculate factorial of each digit
    fact = 1
    for i in range(1, digit + 1):
        fact *= i

    total += fact
    n //= 10

if total == original:
    print(original, "is a Strong Number")
else:
    print(original, "is not a Strong Number")
````

### Problem 7.2: Caesar Cipher (Hard)

Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
- 'A' becomes 'D'
- 'Hello' becomes 'Khoor'

Your program should:
1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

**Hint:** Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

````python
def caesar_cipher(message, shift, mode):
    result = ""

    if mode == "decrypt":
        shift = -shift

    for char in message:
        if char.isalpha():
            start = ord('A') if char.isupper() else ord('a')

            shifted = (ord(char) - start + shift) % 26
            result += chr(start + shifted)
        else:
            result += char

    return result


message = input("Enter your message: ")
shift = int(input("Enter shift value: "))

encrypted = caesar_cipher(message, shift, "encrypt")
print("Encrypted message:", encrypted)

decrypted = caesar_cipher(encrypted, shift, "decrypt")
print("Decrypted message:", decrypted)
````

### Problem 7.3: ATM Machine Simulator (Hard)

Create an ATM machine simulator with the following features:
1. Initial balance: 10000
2. PIN: 1234 (3 attempts allowed)
3. Menu options:
   - Check Balance
   - Withdraw Money (check for sufficient balance and valid amount)
   - Deposit Money
   - Change PIN
   - Exit
4. Keep transaction history (last 5 transactions)
5. Input validation for all operations

**Hint:** Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

````python

balance = 5000
pin = 1234
history = []

# PIN verification
attempts = 0

while attempts < 3:
    entered_pin = int(input("Enter your PIN: "))

    if entered_pin == pin:
        print("Login successful!")
        break
    else:
        attempts += 1
        print("Incorrect PIN. Attempts left:", 3 - attempts)

if attempts == 3:
    print("Your account is locked.")
else:
    # ATM menu
    while True:
        print("\n--- ATM MENU ---")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transaction History")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            print("Current balance:", balance)

        elif choice == "2":
            amount = float(input("Enter deposit amount: "))

            if amount > 0:
                balance += amount
                history.append(f"Deposited: {amount}")
                print("Deposit successful!")
                print("Balance:", balance)
            else:
                print("Amount must be greater than zero.")

        elif choice == "3":
            amount = float(input("Enter withdrawal amount: "))

            if amount <= 0:
                print("Amount must be greater than zero.")
            elif amount > balance:
                print("Insufficient balance.")
            else:
                balance -= amount
                history.append(f"Withdrawn: {amount}")
                print("Withdrawal successful!")
                print("Balance:", balance)

        elif choice == "4":
            print("\n--- Transaction History ---")

            if len(history) == 0:
                print("No transactions yet.")
            else:
                for transaction in history:
                    print(transaction)

        elif choice == "5":
            print("Thank you for using the ATM!")
            break

        else:
            print("Invalid choice. Please try again.")
````

### Problem 10.29: ISBN Validator (Hard)

Validate a 10-digit ISBN number using the check digit algorithm.

**Hint:** ISBN-10: sum = (d1*10 + d2*9 + d3*8 + ... + d10*1). Valid if sum % 11 == 0. Loop with position weight.

````python
isbn = input("Enter 10-digit ISBN: ").strip().upper()

valid_format = (len(isbn) == 10 and isbn[:9].isdigit()
                and (isbn[9].isdigit() or isbn[9] == "X"))

if not valid_format:
    print("Invalid ISBN format")
else:
    total = 0
    weight = 10
    i = 0
    while i < 10:
        value = 10 if isbn[i] == "X" else int(isbn[i])
        total += value * weight
        weight -= 1
        i += 1

    print("Weighted sum:", total)
    if total % 11 == 0:
        print("Valid ISBN")
    else:
        print("Invalid ISBN")
````

### Problem 11.4: Tic-Tac-Toe Game (Hard)

Create a simple two-player Tic-Tac-Toe game.

**Hint:** Use 3x3 list. Loop for turns. Display board, get position, check validity, place X or O. Check win conditions (rows, columns, diagonals) after each move.

````python

# Problem 11.4: Tic-Tac-Toe Game
board = [" "] * 9
wins = [(0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)]


def show_board():
    print()
    for r in range(3):
        print(" " + " | ".join(board[r * 3:r * 3 + 3]))
        if r < 2:
            print("---+---+---")
    print()


def check_winner(player):
    for a, b, c in wins:
        if board[a] == board[b] == board[c] == player:
            return True
    return False


print("Tic-Tac-Toe! Positions are numbered 1-9:")
print(" 1 | 2 | 3")
print("---+---+---")
print(" 4 | 5 | 6")
print("---+---+---")
print(" 7 | 8 | 9")

player = "X"
moves = 0
while moves < 9:
    pos = input(f"Player {player}, choose position (1-9): ")

    if not pos.isdigit() or not 1 <= int(pos) <= 9:
        print("Invalid input! Choose a number between 1 and 9.")
        continue

    pos = int(pos) - 1
    if board[pos] != " ":
        print("That position is already taken!")
        continue

    board[pos] = player
    moves += 1
    show_board()

    if check_winner(player):
        print(f"Player {player} wins!")
        break

    player = "O" if player == "X" else "X"
else:
    print("It's a draw!")
````

### Problem 11.5: Twin Primes (Medium)

Find all twin prime pairs between 1 and n. Twin primes differ by 2.
Example: (3,5), (5,7), (11,13)

**Hint:** Loop through numbers. For each, check if both i and i+2 are prime. If both prime, print pair (i, i+2).

````python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


n = int(input("Enter n: "))

print(f"Twin prime pairs up to {n}:")
for i in range(2, n - 1):
    if is_prime(i) and is_prime(i + 2):
        print((i, i + 2), end=" ")
print()
````
