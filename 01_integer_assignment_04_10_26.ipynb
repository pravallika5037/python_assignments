```python
# Question 1: Calculate the product of first 10 natural numbers
#solution:
n=10
product=1
for i in range(n):
  product=product*(i+1)
print(product)
```

```python
# Question 5: Calculate the sum of digits in number 12345
#solution:
number=12345
sum_of_digit =0
for i in range(number):
  sum_of_digit=sum_of_digit+number%10
  number=number//10
print(sum_of_digit)
```

```python
# Question 6: Check if 97 is a prime number
#solution:
number=97
for i in range(number,number+1):
  if number%i==0:
    print(number,"is not a prime number")
  else:
    print(number,"is a prime number")
    
```

```python
# Question 8: Calculate the average of numbers: 15, 23, 31, 42, 56
#solution:
numbers=[15,23,31,42,56]
average_of_numbers=sum(numbers)/len(numbers)
print(average_of_numbers)

```

```python
# Question 9: Find the greatest common divisor (GCD) of 48 and 36
#solution:
a=48
b=36
common_factors=[]
for i in range(1,min(a,b)+1):
  if a%i==0 and b%i==0:
    common_factors.append(i)
print(max(common_factors))
```

```python
# Question 10: calculate the sum of first 20 odd numbers
#solution:
odd_sum=0
for i in range(1,40):
  # print(i%2!=0)
  if i%2!=0:
    odd_sum=odd_sum+i
print(odd_sum)
```
