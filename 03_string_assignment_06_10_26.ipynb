```python
# Question 41: Check if string contains balanced parentheses: "((()))"
#solution :
text="((()))"
count=0

for char in text:
    if char=="(":
        count=count+1
    elif char==")":
        count=count-1

if count==0:
    print("Balanced parentheses:",True)
else:
    print("Balanced parentheses:",False)
```

```python
# Question 42: Convert "hello world" to Morse code
#solution :
text="hello world"
morse={
    "h":"....",
    "e":".",
    "l":".-..",
    "o":"---",
    "w":".--",
    "r":".-.",
    "d":"-.."
    }

result=""
for char in text:
    if char==" ":
        result=result+"/ "
    else:
        result=result+morse[char]+" "

print("Morse code:",result)
```

```python
# Question 43: Find the longest common substring between "programming" and "grammar"
#solution :
text1="programming"
text2="grammar"
longest=""

for i in range(len(text1)):
    for j in range(i+1,len(text1)+1):
        part=text1[i:j]
        if part in text2 and len(part)>len(longest):
            longest=part

print("Longest common substring:",longest)
```

```python
# Question 47: Check if string is a valid IPv4 address: "192.168.1.1"
#solution :
ip="192.168.1.1"
parts=ip.split(".")

valid=True

if len(parts)!=4:
    valid=False
else:
    for part in parts:
        if not part.isdigit() or int(part)>255:
            valid=False

print("Valid IPv4 address:",valid)
```

```python
# Question 49: Convert "hello world" to ROT13 encoding
#solution :
text="hello world"
result=""

for char in text:
    if char==" ":
        result=result+" "
    else:
        result=result+chr((ord(char)-ord("a")+13)%26+ord("a"))

print("ROT13:",result)
```

```python
# Question 50: Check if string is a valid credit card number: "4532015112830366"
#solution :
number="4532015112830366"
total=0

for i in range(len(number)):
    digit=int(number[i])

    if i%2==0:
        digit=digit*2
        if digit>9:
            digit=digit-9

    total=total+digit

print("Valid credit card number:",total%10==0)
```
